```python
import pandas as pd
import os
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

# Load the CSV file, skipping rows with issues
csv_file_path = "doctors_clinic_timings.csv"
try:
    data = pd.read_csv(csv_file_path, on_bad_lines='skip')
    print("CSV file loaded successfully.")
except Exception as e:
    print(f"Error loading CSV file: {e}")
    exit()

# Fields to include in embeddings
fields_to_include = [
    "Doctor Name", "Education", "Review Details", "Experience", "Rating",
    "Satisfaction", "Clinic Names", "Locations", "Fees", "Clinic Timings",
    "Qualifications", "Services", "Diseases", "Symptoms", "Interests",
    "Contact Number", "Maps Locations"
]

# Ensure all fields exist in the dataset
missing_fields = [field for field in fields_to_include if field not in data.columns]
if missing_fields:
    raise ValueError(f"The following fields are missing in the CSV file: {missing_fields}")

# Display each column being processed
print("Columns being processed:")
for field in fields_to_include:
    print(f"- {field}")

# Concatenate fields into a single text representation for each row
data['Document'] = data[fields_to_include].fillna('').apply(
    lambda row: ' '.join(row.values.astype(str)), axis=1
)
print("Documents created successfully.")


# Initialize the embedding model
embeddings_model = HuggingFaceEmbeddings(
    model_name="dunzhang/stella_en_1.5B_v5",
    model_kwargs={"device": "cpu"}  # Use "cuda" for GPU if available
)
print("Embedding model initialized.")

# Generate FAISS vector store with texts and embedding model
documents = data['Document'].tolist()
vectorstore_path = "vectorstore/doctors_db_faiss2"

if not os.path.exists(os.path.dirname(vectorstore_path)):
    os.makedirs(os.path.dirname(vectorstore_path))

print(f"Generating and saving FAISS vectorstore for {len(documents)} documents...")
vectorstore = FAISS.from_texts(texts=documents, embedding=embeddings_model)
vectorstore.save_local(vectorstore_path)

print(f"Embeddings successfully created and saved at {vectorstore_path}")
```

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv('combined_lawyers_data.csv')

# Group by 'Name' and combine the 'Category' values where duplicates exist
df_grouped = df.groupby('Name').agg({
    'Street Address': 'first',  # Retain the first address (or choose another strategy)
    'Locality': 'first',        # Retain the first locality (or choose another strategy)
    'Phone Number': 'first',    # Retain the first phone number (or choose another strategy)
    'Description': 'first',     # Retain the first description (or choose another strategy)
    'Profile Link': 'first',    # Retain the first profile link (or choose another strategy)
    'Category': lambda x: ', '.join(sorted(set(x)))  # Join categories, remove duplicates
}).reset_index()

# Save the resulting dataframe to a new CSV
df_grouped.to_csv('cleaned_combined_lawyers_data.csv', index=False)

print("CSV file cleaned and saved as 'cleaned_combined_lawyers_data.csv'")
```
